## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Total pages:", len(df))
print("Correlation between search volume and impressions:",
      round(df["search_volume"].corr(df["impressions_90d"]), 3))
print("Median word count by trend:")
print(df.groupby("trend_direction")["word_count"].median())
```

The starter dataset shows that search volume has almost no correlation with impressions, while pages with different trend directions have similar median word counts. This suggests that no single metric is enough to decide which pages need refreshing, making this a suitable machine learning problem.
