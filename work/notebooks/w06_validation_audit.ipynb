# Finding 1

**Paper finding**

The FlyRank research paper reports that growing content was approximately **37.6% longer** and **20% younger** than declining content across the observed dataset. The paper also explains that this is an observational finding rather than proof that longer or newer content causes better performance.

**My methodology question**

One question I would ask is how the labels for "growing" and "declining" content were assigned. I would want to know whether these labels were based on measured traffic changes over a fixed observation period or another process.

I would also ask whether the validation design ensures that related content or clients do not appear across comparison groups. Since the paper presents observational comparisons, I interpret this finding as describing an observed relationship rather than proving causation


# Finding 2

The paper reports that content tends to perform strongest around **61–90** days after publication, declines substantially around **271–365** days, and notes that stronger performance among some very old pages is associated with refreshed content rather than age alone.
flyrank-seo-research-march-2026.pdf

**My methodology question**

I would ask whether content refreshes were separated from naturally aging content during the analysis. If refreshed pages remain grouped with untouched pages, the observed relationship between age and performance could become difficult to interpret.

I would also want to know whether this pattern remains consistent across different content types and clients instead of being driven by a small subset of the dataset.
